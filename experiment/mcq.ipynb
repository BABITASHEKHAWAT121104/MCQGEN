```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    max_new_tokens=100
)
```

```python
response = llm.invoke("What is Python?")
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

prompt1 = PromptTemplate(
    input_variables=["topic"],
    template="""
Generate 3 multiple-choice questions about {topic}.

For each question provide:
Question:
A)
B)
C)
D)
"""
)
```

```python
chain1 = prompt1 | llm
```

```python
prompt2 = PromptTemplate(
    input_variables=["questions"],
    template="""
Here are some multiple-choice questions:

{questions}

For each question:
1. Identify the correct answer.
2. Give a short explanation.

Return the final MCQs in a clean format.
"""
)
```

```python
chain2 = prompt2 | llm
```

```python
chain = chain1 | chain2
```

```python
chain1 = prompt1 | llm

result1 = chain1.invoke({
    "topic": "Python"
})

print(result1)
```

```python
chain2 = prompt2 | llm

result2 = chain2.invoke({
    "questions": result1
})

print(result2)
```

```python
from langchain_core.runnables import RunnableLambda
```

```python
sequential_chain = (
    chain1
    | RunnableLambda(lambda x: {"questions": x})
    | chain2
)
```

```python
result = sequential_chain.invoke({
    "topic": "Python"
})

print(result)
```
